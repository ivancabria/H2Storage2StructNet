```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances

class rbf(object):
    def __init__(self, n_clusters=2):
        # super().__init__(self)
        self.K = n_clusters
        self.label = np.zeros(self.K, dtype=int) - 1
        self.clf = KMeans(n_clusters=self.K, init='k-means++')
        
    def fit(self, X):
        self.clf.fit(X)
        self.centers = self.clf.cluster_centers_
        self.sigma =  pairwise_distances(self.centers).sum(axis=0)/(self.K - 1)
        
    def predict(self, X):
        return self.clf.predict(X)

    def output(self, X):
        return np.exp(-(pairwise_distances(X, self.centers)**2)/
                      (2*(self.sigma**2)))
```
