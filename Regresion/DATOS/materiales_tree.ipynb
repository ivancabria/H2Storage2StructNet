```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier, export_graphviz
import graphviz
```

```python
materiales = pd.read_csv('materiales.csv')
materiales
```

```python
materiales['Class'] = None
```

```python
materiales
```

```python
clave_valor = {"cc2s": 0, "jlu-": 1, "kezb": 2, "mof5": 3, "mof-": 4, "MOF-": 5, "mrts": 6}
```

```python
pd.options.mode.copy_on_write = True
```

```python
for idx in materiales.index:
   materiales.iloc[idx, -1] = clave_valor[materiales['name'][idx][0:4]]
```

```python
materiales
```

```python
X=np.array(materiales.iloc[:, 1:-1])
```

```python
y = np.array(materiales.iloc[:, -1], dtype=int)
```

```python
tree = DecisionTreeClassifier()
```

```python
tree.fit(X, y)
```

```python
export_graphviz(decision_tree=tree, out_file="tree.dot", feature_names=[materiales.columns[1:-1][i][1:-1] for i in range(7)])
```
